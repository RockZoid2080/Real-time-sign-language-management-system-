```python
import cv2 #opencv
import os
import time
import uuid
```

```python
IMAGES_PATH = 'Tensorflow/workspace/images/collectedimages'
```

```python
labels = ['hello','thanks','yes','no','iloveyou']
number_imgs = 15
```

```python
for label in labels:
    os.makedirs('Tensorflow/workspace/images/collectedimages\\'+label, exist_ok=True)  # safer & avoids re-creating dirs
    cap = cv2.VideoCapture(0)
    print('Collecting images for [{}]'.format(label))
    time.sleep(5)
    
    for imgnum in range(number_imgs):
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame")
            break
        
        imgname = os.path.join(IMAGES_PATH, label, label+'.'+'{}.jpg'.format(str(uuid.uuid1())))
        cv2.imwrite(imgname, frame)
        cv2.imshow('frame', frame)
        time.sleep(2)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):  # <- FIXED HERE
            break
    
    cap.release()
    cv2.destroyAllWindows()  # <- closes window after done
```
